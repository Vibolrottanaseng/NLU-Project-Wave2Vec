```python
import json

with open("./annotations/iu_xray_custom.json", "r", encoding="utf-8") as f:
    data = json.load(f)

with open("test_generated_reports.txt", "w", encoding="utf-8") as f:
    for item in data["test"]:
        f.write("both lungs are clear and expanded . heart and mediastinum normal .\n")

print("Created test_generated_reports.txt")
```

```python
with open("test_generated_reports.txt", "r", encoding="utf-8") as f:
    for i in range(5):
        print(f"Generated {i+1}: {f.readline().strip()}")
```

```python
with open("test_generated_reports.txt", "r", encoding="utf-8") as pf, open("test_reference_reports.txt", "r", encoding="utf-8") as rf:
    for i in range(5):
        pred = pf.readline().strip()
        ref = rf.readline().strip()
        print(f"\nSample {i+1}")
        print("Generated   :", pred)
        print("Ground Truth:", ref)
```

```python
!pip install rouge-score -q

from rouge_score import rouge_scorer

scorer = rouge_scorer.RougeScorer(
    ['rouge1', 'rouge2', 'rougeL'],
    use_stemmer=True
)

with open("test_generated_reports.txt", "r", encoding="utf-8") as f:
    preds = [line.strip() for line in f.readlines()]

with open("test_reference_reports.txt", "r", encoding="utf-8") as f:
    refs = [line.strip() for line in f.readlines()]

assert len(preds) == len(refs), f"Mismatch: {len(preds)} predictions vs {len(refs)} references"

r1 = r2 = rL = 0

for pred, ref in zip(preds, refs):
    scores = scorer.score(ref, pred)
    r1 += scores['rouge1'].fmeasure
    r2 += scores['rouge2'].fmeasure
    rL += scores['rougeL'].fmeasure

n = len(preds)

print("ROUGE-1:", r1 / n)
print("ROUGE-2:", r2 / n)
print("ROUGE-L:", rL / n)
```

```python
print("\n| Model | ROUGE-1 | ROUGE-2 | ROUGE-L |")
print("|------|--------:|--------:|--------:|")
print(f"| HC-Gen Adapted | {r1/n:.4f} | {r2/n:.4f} | {rL/n:.4f} |")
```
