```python
# =========================
# PROBLEM LABEL ENCODING
# =========================
def parse_problem_list(x):
    if pd.isna(x):
        return []
    if isinstance(x, list):
        return [str(v).strip().lower() for v in x if str(v).strip()]

    text = str(x).strip()

    try:
        parsed = ast.literal_eval(text)
        if isinstance(parsed, list):
            return [str(v).strip().lower() for v in parsed if str(v).strip()]
    except Exception:
        pass

    if ";" in text:
        return [p.strip().lower() for p in text.split(";") if p.strip()]
    if "," in text:
        return [p.strip().lower() for p in text.split(",") if p.strip()]

    return [text.lower()] if text else []

def encode_problem_vector(problem_list, label_map):
    vec = np.zeros(len(label_map), dtype=np.float32)
    for p in problem_list:
        if p in label_map:
            vec[label_map[p]] = 1.0
    return vec

for df in [train_df, val_df, test_df]:
    df["problem_list_parsed"] = df["problems_list"].apply(parse_problem_list)
    df["problem_vector"] = df["problem_list_parsed"].apply(lambda x: encode_problem_vector(x, label_map))

print(train_df[["problems_list", "problem_list_parsed"]].head())
print("problem_vector shape:", train_df["problem_vector"].iloc[0].shape)
```

```python
import pandas as pd
import json
import ast
import numpy as np
from pathlib import Path
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

from pathlib import Path

BASE = Path("/home/jupyter-st126671/data/iu_xray")
IMG_DIR = BASE / "images"
ANN_PATH = BASE / "annotation_hybrid.json"

train_df = pd.read_csv("/home/jupyter-st126671/train.csv")
val_df = pd.read_csv("/home/jupyter-st126671/val.csv")
test_df = pd.read_csv("/home/jupyter-st126671/test.csv")

with open("/home/jupyter-st126671/problem_label_map.json", "r", encoding="utf-8") as f:
    label_map = json.load(f)

existing_images = {p.name for p in IMG_DIR.iterdir() if p.is_file()}

def has_both_images(row):
    f_ok = str(row["frontal_filename"]) in existing_images
    l_ok = str(row["lateral_filename"]) in existing_images
    return f_ok and l_ok

print("Before filtering")
print("train:", len(train_df))
print("val:", len(val_df))
print("test:", len(test_df))

train_df = train_df[train_df.apply(has_both_images, axis=1)].reset_index(drop=True)
val_df = val_df[val_df.apply(has_both_images, axis=1)].reset_index(drop=True)
test_df = test_df[test_df.apply(has_both_images, axis=1)].reset_index(drop=True)

print("\nAfter filtering")
print("train:", len(train_df))
print("val:", len(val_df))
print("test:", len(test_df))

def parse_problem_list(x):
    if pd.isna(x):
        return []
    if isinstance(x, list):
        return [str(v).strip().lower() for v in x if str(v).strip()]

    text = str(x).strip()

    try:
        parsed = ast.literal_eval(text)
        if isinstance(parsed, list):
            return [str(v).strip().lower() for v in parsed if str(v).strip()]
    except Exception:
        pass

    if ";" in text:
        return [p.strip().lower() for p in text.split(";") if p.strip()]
    if "," in text:
        return [p.strip().lower() for p in text.split(",") if p.strip()]

    return [text.lower()] if text else []

def encode_problem_vector(problem_list, label_map):
    vec = np.zeros(len(label_map), dtype=np.float32)
    for p in problem_list:
        if p in label_map:
            vec[label_map[p]] = 1.0
    return vec

for df in [train_df, val_df, test_df]:
    df["problem_list_parsed"] = df["problems_list"].apply(parse_problem_list)
    df["problem_vector"] = df["problem_list_parsed"].apply(lambda x: encode_problem_vector(x, label_map))

TARGET_COL = "findings_clean"
TOP_K = 3

train_texts = train_df[TARGET_COL].fillna("").astype(str).tolist()
tfidf = TfidfVectorizer(max_features=5000)
train_text_matrix = tfidf.fit_transform(train_texts)
train_problem_matrix = np.vstack(train_df["problem_vector"].values)

def retrieve_similar_cases_from_row(row, top_k=3, w_text=0.5, w_problem=0.5):
    query_prob = row["problem_vector"].reshape(1, -1)
    prob_scores = cosine_similarity(query_prob, train_problem_matrix)[0]

    query_text = str(row[TARGET_COL]) if TARGET_COL in row else ""
    query_text_vec = tfidf.transform([query_text])
    text_scores = cosine_similarity(query_text_vec, train_text_matrix)[0]

    scores = (w_text * text_scores) + (w_problem * prob_scores)

    same_ids = train_df.index[train_df["uid"] == row["uid"]].tolist()
    for idx in same_ids:
        scores[idx] = -1

    top_idx = np.argsort(scores)[-top_k:][::-1]
    retrieved = train_df.iloc[top_idx][["uid", TARGET_COL, "problem_list_parsed"]].copy()
    retrieved["score"] = scores[top_idx]
    return retrieved

def add_retrieval_context(df_split, top_k=3):
    enriched = []
    for _, row in df_split.iterrows():
        retrieved = retrieve_similar_cases_from_row(row, top_k=top_k)

        problem_vector = row["problem_vector"]
        if hasattr(problem_vector, "tolist"):
            problem_vector = problem_vector.tolist()

        example = {
            "id": str(row["uid"]),
            "image_path": [str(row["frontal_filename"]), str(row["lateral_filename"])],
            "report": str(row[TARGET_COL]).strip(),
            "problem_vector": problem_vector,
            "problem_list": row["problem_list_parsed"],
            "retrieved_reports": retrieved[TARGET_COL].astype(str).tolist(),
            "retrieved_problem_lists": retrieved["problem_list_parsed"].tolist(),
            "retrieved_scores": retrieved["score"].tolist()
        }
        enriched.append(example)
    return enriched

annotation_hybrid = {
    "train": add_retrieval_context(train_df, top_k=TOP_K),
    "val": add_retrieval_context(val_df, top_k=TOP_K),
    "test": add_retrieval_context(test_df, top_k=TOP_K),
}

with open(ANN_PATH, "w", encoding="utf-8") as f:
    json.dump(annotation_hybrid, f, indent=2, ensure_ascii=False)

print("Saved cleaned annotation to:", ANN_PATH)
```

```python
# =========================
# FUSION MODULE
# =========================
from pathlib import Path

fusion_file = Path("/home/jupyter-st126671/R2Gen/modules/hybrid_fusion.py")
fusion_code = """
import torch
import torch.nn as nn
import torch.nn.functional as F

class HybridFusion(nn.Module):
    def __init__(self, img_dim, problem_dim, text_dim, hidden_dim):
        super().__init__()
        self.problem_proj = nn.Linear(problem_dim, hidden_dim)
        self.text_proj = nn.Linear(text_dim, hidden_dim)
        self.img_proj = nn.Linear(img_dim, hidden_dim)
        self.attn = nn.Linear(hidden_dim * 3, 3)

    def forward(self, img_feat, problem_vec, retrieved_text_feat):
        img_h = self.img_proj(img_feat)
        prob_h = self.problem_proj(problem_vec)
        text_h = self.text_proj(retrieved_text_feat)

        combined = torch.cat([img_h, prob_h, text_h], dim=-1)
        weights = F.softmax(self.attn(combined), dim=-1)

        fused = (
            weights[:, 0:1] * img_h +
            weights[:, 1:2] * prob_h +
            weights[:, 2:3] * text_h
        )
        return fused, weights
"""
fusion_file.write_text(fusion_code, encoding="utf-8")
print("Created:", fusion_file)
```

```python
from pathlib import Path

hybrid_file = Path("/home/jupyter-st126671/R2Gen/models/hybrid_fusion.py")

hybrid_code = """
import torch
import torch.nn as nn

class HybridFusion(nn.Module):
    def __init__(self, img_dim, problem_dim, text_dim, hidden_dim):
        super(HybridFusion, self).__init__()

        self.problem_proj = nn.Linear(problem_dim, hidden_dim)
        self.text_proj = nn.Linear(text_dim, hidden_dim)
        self.gate = nn.Linear(hidden_dim * 3, hidden_dim)
        self.out = nn.Linear(hidden_dim * 3, hidden_dim)
        self.activation = nn.ReLU()
        self.sigmoid = nn.Sigmoid()

    def forward(self, img_feat, problem_vector, text_feat):
        problem_feat = self.activation(self.problem_proj(problem_vector))
        text_feat = self.activation(self.text_proj(text_feat))

        combined = torch.cat([img_feat, problem_feat, text_feat], dim=-1)
        gate = self.sigmoid(self.gate(combined))
        fused = self.out(combined) * gate
        return fused
"""

hybrid_file.write_text(hybrid_code, encoding="utf-8")
print("Created:", hybrid_file)
```
