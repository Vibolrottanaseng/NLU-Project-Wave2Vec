```python
import kagglehub
import shutil
import os

# Step 1: Download dataset (force clean download)
download_path = kagglehub.dataset_download(
    "raddar/chest-xrays-indiana-university",
    force_download=True
)

# Step 2: Define your target directory
target_path = r"data"

# Step 3: Create target folder
os.makedirs(target_path, exist_ok=True)

# Step 4: Copy files safely
for item in os.listdir(download_path):
    src = os.path.join(download_path, item)
    dst = os.path.join(target_path, item)

    if os.path.exists(dst):
        print(f"⚠️ Skipping (already exists): {item}")
        continue

    if os.path.isdir(src):
        shutil.copytree(src, dst)
    else:
        shutil.copy2(src, dst)

print("✅ Dataset copied to:", target_path)
```
